***Set-Based Program***

Write a program:
***Accept two sets of integers from users***

       *UNION   
       *INTERSECTION  
       *DIFFERENCE


```python
set1 = set(map(int,input("Enter a Set1").split()))
set2 = set(map(int,input("Enter a Set2").split()))
print(set1.union(set2)) #Union 
print(set1.intersection(set2)) # intersection
print(set1.difference(set2))    # Difference
```

***File Handling with Exception Handling***



    **Read a file named data.txt
    **Count number of lines
    **Hande file not found safely

```python
try:
    with open("data.txt",mode="r") as file: # Read a file named data.txt
        
        lines=file.readlines() 
        line_count=len(lines) # count no of lines 
        print(f"Total number of lines:{line_count}")
        
except FileNotFoundError: # Handle file not found safely
    print("Error:File Not Found")
except Exception as e:
    print("An unexpwcted error occured:{e}")
```

***JSON FILE PROGRAM***

Write a Program to:

    * Store books details in a dictinoary
    *Save them into a JSON file
    * Read back and display the content

```python
import json
#1. Store book details in a dictionary
books_data = {
            "book1": {"title": "Science","author": "Krishna","year": 1999},
            "book2": {"title": "Social","author": "Siddu","year": 2000},
            "book3": {"title": "Physics","author": "Kiran","year": 2005}
}
#2.Save them into a  JSON File

with open("books.json","w") as file:
    json.dump(books_data,file)
#3.Read back and display:
with open("books.json","r") as file:
    content=json.load(file)
    print(content)
```
