```python
#same function it performing different actions for different inputs
def add(x,y):
    return x + y

print(add(3,5))
print(add("code","gnan")) # concatination
```

```python
#TYPES OF POLYMORPHISM'S:
#1.Runtime polymorphism----also  called {method overriding}
#2.Compile time polymorphism----also called{Method overloading}
```

```python
#Method overriding:
#person
class Person:
      def __init__(self,name,age):
          self.name=name
          self.age=age
      def show(self):
          return f" Myself {self.name} and iam a {self.age} years old"
#student
class  Student(Person):
    def __init__(self,name,age,std):
        #super().__init__(name,age)
        Person.__init__(self,name,age)
        self.std = std
    def show(self):
          return f" Myself {self.name} and iam studying{self.std} class"
p = Person('krishna',50)
s = Student('sai',15,10)
print(p.show())
print(s.show())
```

```python
#METHOD OVERLOADIING:
class math:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d):
        return a+b+c+d
        #in this only last def  will consider
m=math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
#METHOD OVERLOADIING:
#default aruguments function
class math:
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m=math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
#METHOD OVERLOADIING:
class math:
    def add(self,a,b,c=0,d=0,*args):
        #print(type(args))
        return a+b+c+d
m=math()
print(m.add(1,2,3,4,5,6,7,8,9,12))
print(m.add(1,2,3))
print(m.add(1,2))
```
