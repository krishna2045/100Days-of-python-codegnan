```python
#Inheritence using super():
class User:
    def __init__(self,username:str,email:str,password:str):
        self.username=username
        self.email=email
        self.password=password
    def details(self):
        return f"my name is:{self.username} and my email is:{self.email}"
    def show(self):
        return f"my name is {self.username} and my email is:{self.email}"
        
#Class Admin

class Admin(User):
    def __init__(self,username:str,email:str,password:str,total_users:int):
        super().__init__(username,email,password)
        self.total_users=total_users
    def users(self):
        return f"Iam working as admin and total users are:{self.total_users}"
#Objects creations
#read user details
u1=User(username="krishna",email="gkrishna111911@gmail.com",password="krishna@123")
a1=Admin(username="sai",email="sai1@gmail.com",password="sai@123",total_users=100)

print("user details is:",u1.details())
print(a1.show())
a1.age=30
print("Admin detials is:",a1.details())
```

```python
#1.Singleninheritence
```

```python
#Parent class
class Parent:
    
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
        
    def details(self):
        return f"my self {self.name} and my age is {self.age} years old"
        
#Child class
class Child(Parent):

    def __init__(self, name: str, age: int, std: int):
        self.name = name
        self.age = age
        self.std = std
        
    def details(self):
        
        return f"my child name is {self.name} and age is {self.age} and studying {self.std} class"
        
p=Parent(name="krishna",age=38)
c=Child(name="vamika",age=6,std=3)

print(p.details())
print(c.details())
```

```python
#using super()
# Parent class
class Parent:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
        
    def details(self):
        return f"my self {self.name} and my age is {self.age} years old"

# Child class
class Child(Parent):
    def __init__(self, name: str, age: int, std: int):
        # super() calls the Parent's __init__ to handle name and age
        super().__init__(name, age) 
        self.std = std
        
    def details(self):
        return f"my child name is {self.name} and age is {self.age} and studying {self.std} class"

# Instances
p = Parent(name="krishna", age=38)
c = Child(name="vamika", age=6, std=3)

print(p.details())
print(c.details())
```

```python
#2.Multi-level
```

```python
class Grandfather:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def details(self):
        return f"Myself {self.name}, my age is {self.age} years old."

class Father(Grandfather):
    def __init__(self, name: str, age: int, salary: int):
      
        super().__init__(name, age)
        self.salary = salary
        
    def details(self):
        return f"My name is {self.name} and my salary is {self.salary}."

class Child(Father):
    def __init__(self, name: str, age: int, std: int, salary: int = 0):
        # We pass name, age, and a default salary to the Father class
        super().__init__(name, age, salary)
        self.std = std

    def details(self):
        return f"My child's name is {self.name}, age is {self.age}, and they are in {self.std} class."

# Creating instances
g = Grandfather(name="Venkataiah", age=90)
f = Father(name="Krishna", age=36, salary=49000)
c = Child(name="Vamika", age=6, std=3)

print(g.details())
print(f.details())
print(c.details())
```

```python
class Father:

    def __init__(self,name:str,age:int):
        self.name=name
        self.age=age

    def details(self):
        return f"My Father name is{self.name} and My Father age is{self.age}"

class Mother:

    def __init__(self,name:str,age:int):
        self.name=name
        self.age=age
    def details(self):
        return f"My Mother name is {self.age} and My Mother age is{self.age}"
        
class Child(Father,Mother):
    def __init__(self,name:str,age:int,std:int):
        super().__init__(name,age)
        self.std=std
    def details(self):
        return f"My Name is {self.name}, Iam {self.age} years old and Iam studying in {self.std} class."
f = Father(name="Kankappa",age=46)
m = Mother(name="Parvathi",age=40)
c = Child(name="Krishna",age=20,std=12)

print(f.details())
print(m.details())
print(c.details())
```
