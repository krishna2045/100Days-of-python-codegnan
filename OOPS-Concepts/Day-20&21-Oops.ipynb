```python
# creation of Student class
#Initalization constructor executes whenever an object is created for class
class Student:
    school = "Ravi Teja Model School"  # class attribute

    def __init__(self, name, age, cls):  # constructor
        self.name = name
        self.age = age
        self.cls = cls

    def details(self):
        return f"My name is {self.name}, I am {self.age} years old and I study in class {self.cls}"

# creating objects
s1 = Student("Srinu", 14, 9)
s2 = Student("Sai", 15, 10)

# printing names
print(s1.name)
print(s2.name)

# optional: printing details
print(s1.details())
print(s2.details())
```

```python
#class attribute data shares to all objects
class Student:  # class
    school = "Ravi Teja Model School"  # class attribute

    def __init__(self, name, age, cls):  # constructor
        self.name = name #instance attribute 
        self.age = age
        self.cls = cls

    def details(self): #method
        return f"My name is {self.name}, I am {self.age} years old and I study in class {self.cls}"

# creating objects
s1 = Student("Srinu", 14, 9)
s2 = Student("Sai", 15, 10)

# printing names
print(s1.school)
print(s2.school)
```

```python
#Update instance and class attribute outside the class
class Student:
    school = "Ravi Teja Model School"  # class attribute

    def __init__(self, name, age, cls):  # constructor
        self.name = name
        self.age = age
        self.cls = cls

    def details(self):
        return f"My name is {self.name}, I am {self.age} years old and I study in class {self.cls}"

    def detail(self,percentage):
        return f" Myself{self.name} and my percentage is{percentage}"

# creating objects
s1 = Student("Srinu", 14, 9)
s2 = Student("Sai", 15, 10)
s1.name ="srinu"
print(s1.name)
s1.school='ABc school'
print(s1.school)
print(s2.school)
```
