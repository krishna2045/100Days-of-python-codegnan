```python
try:
    print(a)
except:
    print("Error occured")
else:
    print("Error not occured")
finally:
    print("Alawys")
```

```python
try:
    a=10
    b=0
    c=a/b
    print(c)
except:
    print("Error is occured")
```

```python
try:
    print('a'+10)
except Exception as krishna:
    print("Error occured",krishna)
```

```python
try:
    print('a'+10)
except TypeError:
    print("Type Error")
except NameError:
    print("Name Error")
```

```python
try:
    print('k'+10)
    print(s)
```

```python
#Multiple Error:using nested try
try:
    print('s'+10)
except:
    print("Outer")
    try:
        print(s)
    except:
        print("inner")
```
